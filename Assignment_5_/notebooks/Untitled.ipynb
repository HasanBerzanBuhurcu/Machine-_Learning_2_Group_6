```python
import yfinance as yf
import numpy as np
import pandas as pd

TICKER = "^GSPC"   # or "^GDAXI" for DAX
START, END = "2015-01-01", "2025-01-01"

px = yf.download(TICKER, start=START, end=END, auto_adjust=True)["Close"]
ret = (np.log(px / px.shift(1)) * 100).dropna()
ret.name = "ret"
ret.to_csv("data/returns.csv")

print(ret.shape, ret.index.min(), ret.index.max())
print(ret.describe())
```
